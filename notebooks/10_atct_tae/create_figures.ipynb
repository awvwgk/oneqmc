```python
import re
from oneqmc.analysis.energy import HARTREE_TO_KCAL
from uncertainties import ufloat_fromstr
from uncertainties.unumpy import nominal_values, std_devs
import matplotlib.pyplot as plt
import pandas as pd
```

```python
results_df = pd.read_csv("../../experiment_results/10_atct_tae/processed_results.csv")
results_df["orbformer_tae_hartree"] = results_df["orbformer_tae_hartree"].apply(ufloat_fromstr)
results_df["error_hartree"] = results_df["error_hartree"].apply(ufloat_fromstr)
```

```python
def digit_to_subscript(formula: str) -> str:
    # Replace all digits with a LaTeX style subscript
    return re.sub(r"(\d)", r"$_{\1}$", formula)

fig, ax = plt.subplots(figsize=(6, 3.5))
length = len(results_df)
ax.errorbar(range(len(results_df)), HARTREE_TO_KCAL * nominal_values(results_df["error_hartree"]), yerr=HARTREE_TO_KCAL * std_devs(results_df["error_hartree"]), fmt="o")
ax.fill_between([-1, length], -1, 1, color="gray", alpha=0.2)
ax.axhline(0, color="gray", linestyle="--")
ax.set_xlim(-0.7, length - 0.3)
ax.set_xticks(range(len(results_df)))
ax.set_xticklabels(results_df["mol_name"].apply(digit_to_subscript), rotation=90)
ax.set_ylabel(r"Diff. to experimental $\Delta H^\circ_f(0 \, \text{K})$ (kcal/mol)")
print(f'{nominal_values(results_df["error_hartree"].abs()).mean() * HARTREE_TO_KCAL:.2f} kcal/mol MAE')
fig.savefig("atct_tae_error.pdf", bbox_inches="tight", dpi=300)
```
